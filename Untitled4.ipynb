```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
import kagglehub
from kagglehub import KaggleDatasetAdapter
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np


# 1. Download the dataset files to your local cache
path = kagglehub.dataset_download("shivamb/netflix-shows")
print("Path to dataset files:", path)

# 2. Read the CSV file directly from that path into pandas
df = pd.read_csv(f"{path}/netflix_titles.csv")

# 3. Save a copy locally in your current project folder
df.to_csv("netflix.csv", index=False)
print("Dataset saved locally as netflix_titles_local.csv")
```

```python
df.info()
```

```python
missing_values_percentage = ((df.isnull().sum() / len(df)) * 100).round(2)
print("Percentage of missing values in each column:")
print(missing_values_percentage)
```

```python
df['director'].value_counts()
```

```python
temp_df = df.dropna(subset=['director'])

# Compare the distribution of top directors before and after dropping other missing data
top_directors_original = df['director'].value_counts(normalize=True).head(10)
top_directors_cca = temp_df['director'].value_counts(normalize=True).head(10)

comparison_df = pd.DataFrame({
    'With Null Values (Original)': top_directors_original,
    'Without Null Values (CCA)': top_directors_cca
})

# Plot a comparison of the proportions of top directors
comparison_df.plot(kind='bar', figsize=(10, 6))
plt.title('Proportion of Top 10 Directors: Original vs. CCA')
plt.ylabel('Relative Frequency (Proportion)')
plt.xlabel('Director')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df.dropna(subset=['director'], inplace=True)
df.dropna(subset=['country'], inplace=True)
df.dropna(subset=['date_added'], inplace=True)
df.dropna(subset=['rating'], inplace=True)
df.dropna(subset=['duration'], inplace=True)
df.dropna(subset=['cast'], inplace=True)
```

```python
df['duration'].head(30)
```

```python
units = df['duration'].str.extract(r'\d+\s+([a-zA-Z]+)')[0]
print(units.value_counts())
```

```python
movies = df[df['type'] == 'Movie'].copy()

movies['duration'].describe()
```

```python
tv_shows = df[df['type'] == 'TV Show'].copy()
tv_shows['duration'].describe()
```

```python
tv_shows['date_added'].head()
```

```python
tv_shows['date_added'] = pd.to_datetime(
    tv_shows['date_added'].str.strip(),
    format='%B %d, %Y'
)
movies['date_added'] = pd.to_datetime(
    movies['date_added'].str.strip(),
    format='%B %d, %Y'
)
```

```python
print(tv_shows['date_added'].head())
print(movies['date_added'].head())
```

```python
# Save the cleaned combined data, movies, and TV shows to new CSV files
df.to_csv('netflix_cleaned.csv', index=False)
movies.to_csv('netflix_movies_cleaned.csv', index=False)
tv_shows.to_csv('netflix_shows_cleaned.csv', index=False)

print("Saved the following files:")
print("- netflix_cleaned.csv (Combined cleaned cases)")
print("- netflix_movies_cleaned.csv (Cleaned Movies only with converted dates)")
print("- netflix_shows_cleaned.csv (Cleaned TV Shows only with converted dates)")
```
